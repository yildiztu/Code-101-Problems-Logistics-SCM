# 40. The Port Capacity and Expansion Timing Problem
## Tier 10: Proactive Demand and Market Shaping

### Goal
This case explores how a port authority might influence when and how customers use port services, rather than responding only to demand after it appears. The notebook summarizes market segments, compares five proposed initiatives, evaluates a specified three-initiative portfolio, and illustrates the portfolio's potential financial and operational effects.

### Inputs, outputs, and scope
The inputs are illustrative market-segment estimates and initiative assumptions. The selected portfolio is specified directly in the code; it is not produced by an optimization algorithm. The outputs are arithmetic summaries of those inputs, a set of explicitly labeled scenario assumptions, and visualizations. The notebook therefore demonstrates a market-shaping assessment, not an empirically validated forecast or a portfolio optimization model.

The model assumes initiative effects can be added. This is a simplifying assumption: in practice, initiatives may overlap, reinforce one another, or affect different parts of the market. The financial illustration assumes a $50 million annual revenue baseline and treats each initiative's revenue impact as a percentage of that baseline. It does not include operating costs, taxes, financing, inflation, discounting, implementation risk, or revenue interactions. The quarterly utilization profile and other operational scenario figures are illustrative inputs, not effects calculated from the initiative data.

### Market-shaping approach
The initiatives represent six broad capabilities: dynamic pricing, supply-chain coordination, behavioral nudging, ecosystem development, predictive market making, and strategic market influence. In this notebook, the first five are represented by quantified initiative assumptions. Strategic market influence is discussed as context rather than assigned a separate numerical initiative.

### What to examine
- How market segments differ in size, price elasticity, service sensitivity, and growth potential.
- How the specified portfolio's investment and additive impact totals are calculated.
- How the financial results depend on the revenue-baseline assumption.
- Which scenario statements are calculated from the inputs and which are assumptions requiring external evidence.

### Illustrative market scenario
The scenario describes peak utilization of 94% in Q4, off-peak utilization of 67% in Q1–Q2, and average vessel-arrival deviation of ±18 hours. It also lists peak-period pricing 45% higher and off-peak pricing 35% lower, a partnership with 12 major shippers, a $67 million specialized-facilities investment, and market concentration changing from 78% to 58%. These are scenario statements, not results derived by the calculations below. Their real-world interpretation would require definitions, data, and evidence not supplied in this notebook.

### Interpretation and limitations
A large stated demand shift or congestion reduction is not, by itself, proof of improved welfare, lower total costs, or a feasible operating plan. Pricing and behavioral interventions also raise questions about fairness, transparency, competition, and the distribution of benefits and costs among shippers and communities. Those questions require stakeholder evidence and policy analysis beyond this numerical illustration.

The purpose of the case is to make the assumptions and arithmetic visible. The results should be read as conditional illustrations: they follow only if the stated inputs and additive-effect assumption are accepted.

## Quantitative model

### Segment shares
Let \(S\) be the set of market segments, and let \(v_s\) denote the estimated annual volume, in TEU, for segment \(s\). The total market size and each segment's share are

$$
V = \sum_{s \in S} v_s,
\qquad
q_s = \frac{v_s}{V} \times 100\%.
$$

Here, \(V\) is the sum of the listed segment volumes and \(q_s\) is a descriptive share of that total. In the code, these quantities are `total_market_size` and `market_share`.

### Initiative totals and financial illustration
Let \(I\) be the set of initiatives and \(P \subseteq I\) the specified portfolio. For initiative \(i\), let \(c_i\) be its investment in millions of dollars, \(d_i\) its demand-shift assumption in percentage points, \(g_i\) its congestion-reduction assumption in percentage points, and \(r_i\) its revenue-impact assumption in percentage points. Under the additive-effect assumption, the portfolio totals are

$$
C_P = \sum_{i \in P} c_i,
\qquad
D_P = \sum_{i \in P} d_i,
\qquad
G_P = \sum_{i \in P} g_i,
\qquad
R_P = \sum_{i \in P} r_i.
$$

The addition is a modeling assumption, not a general law: it does not account for overlap, saturation, or interactions between initiatives. The percentage-point totals should not be interpreted as probabilities or guaranteed changes.

For an annual baseline revenue \(B\), expressed in millions of dollars, the illustrative annual revenue gain and gross revenue-to-investment multiple are

$$
\Delta A_P = \frac{R_P}{100} B,
\qquad
M_P = \frac{\Delta A_P}{C_P}.
$$

The same calculation applies to one initiative by replacing \(R_P\) and \(C_P\) with \(r_i\) and \(c_i\). A multiple of this kind compares annual revenue gain with investment; it is not net ROI and does not subtract investment or account for the timing of cash flows. A simple payback period, when annual gain is positive, is \(C_P / \Delta A_P\) years. A five-year cumulative net benefit without discounting is \(5\Delta A_P - C_P\); it is not a net present value.

In the code, the portfolio quantities are `total_investment`, `total_demand_shift`, `total_congestion_reduction`, and `total_revenue_impact`. `base_annual_revenue`, `annual_revenue_gain`, `overall_roi`, `payback_period`, and `five_year_cumulative_net_benefit` implement the financial definitions. No decision variables, budget constraint, or objective function are defined, so this case does not establish that the specified portfolio is optimal.

```python
import random
import warnings

import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

warnings.filterwarnings("ignore")

# Fixed seeds keep any random operations reproducible if they are added to this case.
np.random.seed(42)
random.seed(42)

plt.style.use("default")
sns.set_palette("husl")

print("Proactive demand and market-shaping analysis initialized.")
```

## 1. Describe the market and initiative assumptions

A market segment is a group of customers represented here by its estimated volume and three behavioral assumptions. Price elasticity describes the percentage response in demand associated with a percentage change in price; a negative value indicates that demand is assumed to fall as price rises. Service sensitivity and growth potential are scenario descriptors. The notebook reports them but does not use them to forecast demand or select initiatives.

Each initiative has an assumed investment and three percentage-point impacts. These inputs are treated as independent and additive when the portfolio totals are calculated. The direct selection below is a specified example portfolio, not the output of a search or optimization procedure.

```python
print("\n" + "=" * 80)
print("PROACTIVE DEMAND AND MARKET SHAPING: PORT CAPACITY PLANNING")
print("=" * 80)

market_segments = [
    {
        "name": "Major Shipping Lines",
        "size": 500000,
        "price_elasticity": -0.3,
        "service_sensitivity": 0.8,
        "growth_potential": 0.03,
    },
    {
        "name": "Regional Carriers",
        "size": 200000,
        "price_elasticity": -0.5,
        "service_sensitivity": 0.9,
        "growth_potential": 0.05,
    },
    {
        "name": "Niche Specialists",
        "size": 80000,
        "price_elasticity": -0.2,
        "service_sensitivity": 0.7,
        "growth_potential": 0.08,
    },
    {
        "name": "Bulk Cargo Operators",
        "size": 300000,
        "price_elasticity": -0.4,
        "service_sensitivity": 0.6,
        "growth_potential": 0.02,
    },
    {
        "name": "Emerging Markets",
        "size": 120000,
        "price_elasticity": -0.6,
        "service_sensitivity": 0.9,
        "growth_potential": 0.12,
    },
]

print("\nMARKET SEGMENTS")
total_market_size = sum(segment["size"] for segment in market_segments)

for segment in market_segments:
    market_share = segment["size"] / total_market_size * 100
    print(f"   {segment['name']}:")
    print(f"     Size: {segment['size']:,} TEU ({market_share:.1f}% of listed volume)")
    print(f"     Price elasticity: {segment['price_elasticity']:.2f}")
    print(f"     Service sensitivity: {segment['service_sensitivity']:.2f}")
    print(f"     Growth potential: {segment['growth_potential']:.1%} per year")

initiatives = [
    {
        "name": "Dynamic Pricing System",
        "investment": 15.0,
        "demand_shift": 12.0,
        "congestion_reduction": 18.0,
        "revenue_impact": 8.5,
    },
    {
        "name": "Supply Chain Integration",
        "investment": 28.0,
        "demand_shift": 18.0,
        "congestion_reduction": 25.0,
        "revenue_impact": 12.3,
    },
    {
        "name": "Behavioral Nudging",
        "investment": 8.0,
        "demand_shift": 8.0,
        "congestion_reduction": 12.0,
        "revenue_impact": 4.2,
    },
    {
        "name": "Ecosystem Development",
        "investment": 67.0,
        "demand_shift": 35.0,
        "congestion_reduction": 42.0,
        "revenue_impact": 28.7,
    },
    {
        "name": "Predictive Market Making",
        "investment": 22.0,
        "demand_shift": 15.0,
        "congestion_reduction": 20.0,
        "revenue_impact": 10.8,
    },
]

base_annual_revenue = 50.0

print("\nMARKET-SHAPING INITIATIVE ASSUMPTIONS")
for initiative in initiatives:
    annual_gain = initiative["revenue_impact"] / 100 * base_annual_revenue
    revenue_to_investment = annual_gain / initiative["investment"]
    print(f"   {initiative['name']}:")
    print(f"     Investment: ${initiative['investment']:.1f}M")
    print(f"     Assumed demand shift: {initiative['demand_shift']:.1f} percentage points")
    print(
        "     Assumed congestion reduction: "
        f"{initiative['congestion_reduction']:.1f} percentage points"
    )
    print(f"     Assumed revenue impact: {initiative['revenue_impact']:.1f}%")
    print(f"     Illustrative annual revenue gain: ${annual_gain:.2f}M")
    print(
        "     Gross annual revenue-gain-to-investment multiple: "
        f"{revenue_to_investment:.2f}x"
    )

# This specified portfolio is used to demonstrate the additive calculations;
# no optimization model or portfolio search is performed in this notebook.
selected_initiative_names = {
    "Supply Chain Integration",
    "Ecosystem Development",
    "Predictive Market Making",
}
best_portfolio = [
    initiative
    for initiative in initiatives
    if initiative["name"] in selected_initiative_names
]

assert len(best_portfolio) == len(selected_initiative_names)

total_investment = sum(item["investment"] for item in best_portfolio)
total_demand_shift = sum(item["demand_shift"] for item in best_portfolio)
total_congestion_reduction = sum(
    item["congestion_reduction"] for item in best_portfolio
)
total_revenue_impact = sum(item["revenue_impact"] for item in best_portfolio)
annual_revenue_gain = total_revenue_impact / 100 * base_annual_revenue
overall_roi = annual_revenue_gain / total_investment
payback_period = total_investment / annual_revenue_gain
five_year_cumulative_net_benefit = 5 * annual_revenue_gain - total_investment

assert total_investment == sum(item["investment"] for item in best_portfolio)
assert np.isclose(
    annual_revenue_gain,
    total_revenue_impact / 100 * base_annual_revenue,
)

print("\nSPECIFIED ILLUSTRATIVE PORTFOLIO")
print(f"   Selected initiatives: {len(best_portfolio)}")
print(f"   Total investment: ${total_investment:.1f}M")
print(f"   Additive demand-shift assumption: {total_demand_shift:.1f} percentage points")
print(
    "   Additive congestion-reduction assumption: "
    f"{total_congestion_reduction:.1f} percentage points"
)
print(f"   Additive revenue-impact assumption: {total_revenue_impact:.1f}%")
print(f"   Assumed annual revenue baseline: ${base_annual_revenue:.1f}M")
print(f"   Illustrative annual revenue gain: ${annual_revenue_gain:.2f}M")
print(f"   Gross annual revenue-gain-to-investment multiple: {overall_roi:.2f}x")
print(f"   Simple payback period: {payback_period:.2f} years")
print(
    "   Five-year cumulative net benefit, without discounting: "
    f"${five_year_cumulative_net_benefit:.2f}M"
)

print("\nPORTFOLIO MEMBERS")
for initiative in best_portfolio:
    print(f"   - {initiative['name']} (${initiative['investment']:.1f}M)")
```

## 2. Read the portfolio and financial results

The code adds the assumptions for the three named initiatives. For example, `total_investment` is the sum of their investments, while `total_revenue_impact` is the sum of their stated revenue-impact percentages. The annual revenue gain converts that percentage to a fraction and applies it to the assumed $50 million annual baseline.

The reported multiple is annual revenue gain divided by investment. It is a gross comparison, not a complete profitability measure. The simple payback calculation assumes the annual gain is realized in full and remains constant. The five-year cumulative net benefit subtracts the initial investment from five years of the same annual gain, without discounting. These calculations do not establish that the portfolio is financially feasible or optimal.

```python
print("\n" + "=" * 60)
print("ILLUSTRATIVE MARKET AND IMPLEMENTATION SCENARIO")
print("=" * 60)

# These scenario values are stated assumptions; they are not estimated by the
# initiative arithmetic or inferred from the segment data.
scenario_assumptions = {
    "baseline_utilization": 0.78,
    "baseline_schedule_deviation_hours": 18,
    "baseline_market_concentration_percent": 78,
    "scenario_schedule_deviation_hours": 12,
    "scenario_market_concentration_percent": 58,
    "market_share_change_percent": 20.4,
    "brand_value_change_percent": 40.8,
    "customer_loyalty_change_percent": 13.6,
    "network_effect_multiplier": 1.34,
    "implementation_months": 24,
}

print("\nScenario assumptions (not model outputs):")
print(
    "   Baseline port utilization: "
    f"{scenario_assumptions['baseline_utilization']:.1%}"
)
print("   Peak-period congestion: described qualitatively as high")
print(
    "   Baseline vessel-arrival deviation: "
    f"±{scenario_assumptions['baseline_schedule_deviation_hours']} hours"
)
print(
    "   Baseline market concentration: "
    f"{scenario_assumptions['baseline_market_concentration_percent']}%"
)
print("\nIllustrative scenario values after market-shaping interventions:")
print(
    "   Vessel-arrival deviation: "
    f"±{scenario_assumptions['scenario_schedule_deviation_hours']} hours"
)
print(
    "   Market concentration: "
    f"{scenario_assumptions['scenario_market_concentration_percent']}%"
)
print(
    "   Market-share change: "
    f"{scenario_assumptions['market_share_change_percent']:.1f}%"
)
print(
    "   Brand-value change: "
    f"{scenario_assumptions['brand_value_change_percent']:.1f}%"
)
print(
    "   Customer-loyalty change: "
    f"{scenario_assumptions['customer_loyalty_change_percent']:.1f}%"
)
print(
    "   Network-effect multiplier: "
    f"{scenario_assumptions['network_effect_multiplier']:.2f}x"
)

print("\nFinancial calculations from the stated portfolio assumptions:")
print(f"   Total investment: ${total_investment:.1f}M")
print(f"   Annual revenue gain: ${annual_revenue_gain:.2f}M")
print(f"   Simple payback period: {payback_period:.2f} years")
print(
    "   Five-year cumulative net benefit, without discounting: "
    f"${five_year_cumulative_net_benefit:.2f}M"
)
print(f"   Gross annual revenue-gain-to-investment multiple: {overall_roi:.2f}x")

print("\nPlanning considerations (not calculated recommendations):")
print("   - Validate demand responses and congestion effects with operational data.")
print("   - Assess distributional, competition, privacy, and transparency implications.")
print("   - Define measurable milestones and review assumptions as evidence changes.")
print(
    "   - The scenario assumes a "
    f"{scenario_assumptions['implementation_months']}-month implementation horizon."
)

print("\nInterpretation:")
print(
    "   The portfolio arithmetic follows from the supplied assumptions; "
    "the operational scenario values require independent validation."
)
print("=" * 60)
```

## 3. Visualize the inputs and scenario profile

The plots compare the supplied segment and initiative assumptions. The quarterly utilization chart uses a separate illustrative profile; it is not calculated from the initiative impacts. A star marks an initiative in the specified portfolio, while the legend distinguishes portfolio membership. These visual summaries help inspect the inputs but do not test whether the scenario assumptions are causally supported.

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle(
    "Illustrative Market-Shaping Inputs and Scenario Profile",
    fontsize=16,
    fontweight="bold",
)

# Segment volumes are displayed in TEU; bar colors distinguish the categories.
segment_sizes = [segment["size"] for segment in market_segments]
segment_names = [segment["name"] for segment in market_segments]
colors = ["skyblue", "lightgreen", "salmon", "gold", "mediumpurple"]

segment_bars = axes[0, 0].bar(segment_names, segment_sizes, color=colors)
axes[0, 0].set_title("Estimated Market Segment Volumes")
axes[0, 0].set_ylabel("Volume (TEU)")
axes[0, 0].tick_params(axis="x", rotation=35)
axes[0, 0].grid(True, axis="y", alpha=0.3)

for bar, size in zip(segment_bars, segment_sizes):
    axes[0, 0].text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{size:,}",
        ha="center",
        va="bottom",
        fontsize=8,
    )

# The star identifies membership in the specified portfolio, not an optimized choice.
investments = [item["investment"] for item in initiatives]
demand_shifts = [item["demand_shift"] for item in initiatives]
initiative_names = [item["name"] for item in initiatives]
selected_names = {item["name"] for item in best_portfolio}

axes[0, 1].scatter(
    investments,
    demand_shifts,
    s=90,
    color="steelblue",
    alpha=0.75,
    label="Other listed initiative",
)
selected_points = [
    item for item in initiatives if item["name"] in selected_names
]
axes[0, 1].scatter(
    [item["investment"] for item in selected_points],
    [item["demand_shift"] for item in selected_points],
    s=220,
    color="red",
    marker="*",
    edgecolors="black",
    linewidths=0.8,
    label="Specified portfolio member",
)
axes[0, 1].set_xlabel("Investment assumption ($M)")
axes[0, 1].set_ylabel("Demand-shift assumption (percentage points)")
axes[0, 1].set_title("Investment and Assumed Demand Shift")
axes[0, 1].grid(True, alpha=0.3)
axes[0, 1].legend()

for item in initiatives:
    axes[0, 1].annotate(
        item["name"],
        (item["investment"], item["demand_shift"]),
        xytext=(5, 5),
        textcoords="offset points",
        fontsize=7,
    )

# These quarterly values are an illustrative profile, independent of the portfolio sums.
periods = ["Q1", "Q2", "Q3", "Q4"]
baseline_quarterly_utilization = [0.67, 0.72, 0.85, 0.94]
scenario_quarterly_utilization = [0.75, 0.78, 0.82, 0.88]
x = np.arange(len(periods))
width = 0.35

baseline_bars = axes[1, 0].bar(
    x - width / 2,
    baseline_quarterly_utilization,
    width,
    label="Illustrative baseline",
    color="lightcoral",
)
scenario_bars = axes[1, 0].bar(
    x + width / 2,
    scenario_quarterly_utilization,
    width,
    label="Illustrative scenario",
    color="lightgreen",
)
axes[1, 0].set_xlabel("Quarter")
axes[1, 0].set_ylabel("Port utilization (fraction)")
axes[1, 0].set_title("Illustrative Quarterly Utilization Profile")
axes[1, 0].set_xticks(x)
axes[1, 0].set_xticklabels(periods)
axes[1, 0].set_ylim(0, 1.05)
axes[1, 0].legend()
axes[1, 0].grid(True, axis="y", alpha=0.3)

for bar_group in [baseline_bars, scenario_bars]:
    for bar in bar_group:
        height = bar.get_height()
        axes[1, 0].text(
            bar.get_x() + bar.get_width() / 2,
            height + 0.015,
            f"{height:.0%}",
            ha="center",
            va="bottom",
            fontsize=8,
        )

# Each bar is a gross annual revenue-gain-to-investment multiple under the stated baseline.
initiative_multiples = [
    (item["revenue_impact"] / 100 * base_annual_revenue) / item["investment"]
    for item in initiatives
]
initiative_labels = [item["name"] for item in initiatives]
bar_colors = [
    "red" if item["name"] in selected_names else "steelblue"
    for item in initiatives
]
roi_bars = axes[1, 1].barh(
    initiative_labels,
    initiative_multiples,
    color=bar_colors,
)
axes[1, 1].set_xlabel("Gross annual revenue-gain-to-investment multiple (x)")
axes[1, 1].set_title("Illustrative Initiative Financial Ratios")
axes[1, 1].grid(True, axis="x", alpha=0.3)

for bar, multiple in zip(roi_bars, initiative_multiples):
    axes[1, 1].text(
        bar.get_width() + 0.01,
        bar.get_y() + bar.get_height() / 2,
        f"{multiple:.2f}x",
        ha="left",
        va="center",
        fontsize=8,
    )

plt.tight_layout()
plt.show()

print("Visualizations display the listed assumptions and illustrative quarterly profile.")
```
