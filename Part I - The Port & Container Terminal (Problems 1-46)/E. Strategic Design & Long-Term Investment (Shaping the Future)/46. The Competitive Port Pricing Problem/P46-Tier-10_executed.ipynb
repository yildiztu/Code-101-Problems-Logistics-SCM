# 46. The Competitive Port Pricing Problem

## Tier 10: Proactive Demand and Market Shaping

### Goal

This notebook presents a structured framework for thinking about how a port might shape demand through strategic interventions. It identifies example market opportunities, associates each opportunity with a possible intervention, and organizes the information needed to assess those interventions.

The notebook does **not** estimate demand responses, optimize prices, or calculate realized market outcomes. The case figures and opportunity attributes are illustrative inputs; they are not results produced by the code.

### Inputs and outputs

- **Inputs:** Port capacities and costs, shipping-line characteristics, market parameters, and illustrative opportunity descriptions.
- **Outputs:** Opportunity records, intervention designs, a report that distinguishes specified information from unavailable impact estimates, and visual summaries of the inputs.
- **Method:** A rule-based mapping from each opportunity type to a proposed intervention. No optimization algorithm or calibrated demand model is specified in this example.

### Scope and modeling assumptions

The example treats the supplied market and opportunity attributes as planning assumptions. Customer behavior profiles are descriptive segment profiles, not estimated customer responses. The price-sensitivity parameter is stored as market information but is not used to calculate demand or a price. Investment costs, intervention effects, and realized outcomes are not supplied, so the notebook leaves them unquantified rather than presenting invented estimates. Capacity, costs, and demand are displayed for context; they are not used to solve an allocation or pricing model.

### Strategic context

A proactive market-shaping strategy can include volume incentives for emerging trade lanes, premium services for high-value customers, and partnerships with logistics providers. The accompanying case description reports growth and positioning figures for that example. Those reported figures are contextual benchmarks only; this notebook does not reproduce them as calculated or achieved results.

### Learning sequence

1. Define the entities and market information represented in the example.
2. Identify opportunities from the supplied opportunity descriptions.
3. Map opportunities to intervention designs using explicit rules.
4. Separate known planning inputs from impact measures that require additional evidence.
5. Visualize the opportunity and intervention information without implying unmeasured outcomes.

```python
from dataclasses import dataclass
from enum import Enum
from typing import Dict, List, Optional, Tuple
from collections import defaultdict

import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

plt.style.use("seaborn-v0_8")
sns.set_palette("husl")
```

```python
@dataclass
class Port:
    """A port with capacity, shipping-line costs, and a strategic position."""

    id: int
    name: str
    capacity: float  # Maximum capacity in TEU
    costs: List[float]  # Cost per shipping line
    strategic_position: str


@dataclass
class ShippingLine:
    """A shipping line with demand and customer-segment characteristics."""

    id: int
    name: str
    demand: float  # Demand in TEU
    priority: int  # 1 = high, 2 = medium, 3 = low
    segment: str
    price_elasticity: float
    growth_potential: float
    loyalty_score: float


@dataclass
class MarketParameters:
    """Market-wide assumptions supplied with the example."""

    price_sensitivity: float
    min_price: float
    max_price: float
    market_growth_rate: float
    competitive_intensity: float


class MarketInterventionType(Enum):
    """Intervention categories used by the opportunity-to-action mapping."""

    VOLUME_DISCOUNT = "volume_discount"
    PREMIUM_SERVICE = "premium_service"
    PARTNERSHIP_PROGRAM = "partnership_program"
    LOYALTY_INCENTIVE = "loyalty_incentive"
    NEW_MARKET_DEVELOPMENT = "new_market_development"
    COMPETITIVE_POSITIONING = "competitive_positioning"
    DEMAND_STIMULATION = "demand_stimulation"
    BEHAVIORAL_NUDGE = "behavioral_nudge"


@dataclass
class MarketIntervention:
    """A proposed action, with unknown costs and impacts left unspecified."""

    intervention_type: MarketInterventionType
    target_segments: List[str]
    pricing_strategy: str
    expected_impact: Dict[str, float]
    duration_weeks: int
    investment_required: Optional[float]
    success_metrics: List[str]
    risk_factors: List[str]


@dataclass
class CustomerBehavior:
    """A descriptive profile for a customer segment."""

    segment: str
    price_sensitivity: float
    service_preference: float
    loyalty_drivers: List[str]
    switching_costs: float
    growth_indicators: List[str]


@dataclass
class MarketOpportunity:
    """An illustrative opportunity record used for planning discussion."""

    opportunity_type: str
    target_customers: List[str]
    market_size: float
    growth_rate: float
    profit_potential: float
    competitive_advantage: str
    required_capabilities: List[str]
    time_to_market: int  # Weeks
    confidence_level: float
```

```python
class MarketShapingEngine:
    """Organize market information and produce rule-based intervention designs."""

    def __init__(
        self,
        ports: List[Port],
        shipping_lines: List[ShippingLine],
        market_params: MarketParameters,
    ) -> None:
        self.ports = ports
        self.shipping_lines = shipping_lines
        self.market_params = market_params
        self.num_ports = len(ports)
        self.num_shipping_lines = len(shipping_lines)

        self.market_trends = defaultdict(list)
        self.customer_behaviors: Dict[int, CustomerBehavior] = {}
        self.competitive_landscape = {}
        self.active_interventions: List[MarketIntervention] = []
        self.intervention_history: List[MarketIntervention] = []
        self.intervention_performance = defaultdict(list)
        self.identified_opportunities: List[MarketOpportunity] = []
        self.opportunity_pipeline: List[MarketOpportunity] = []
        self.market_shaping_metrics = {
            "demand_shift": [],
            "behavior_change": [],
            "market_growth": [],
            "competitive_position": [],
            "customer_loyalty": [],
            "new_market_penetration": [],
        }

        self._initialize_customer_behaviors()
        self._initialize_competitive_landscape()

    def _initialize_customer_behaviors(self) -> None:
        """Attach descriptive segment profiles to the supplied shipping lines."""

        behavior_profiles = {
            "premium": CustomerBehavior(
                segment="premium",
                price_sensitivity=0.3,
                service_preference=0.9,
                loyalty_drivers=["service_quality", "reliability", "exclusive_access"],
                switching_costs=0.8,
                growth_indicators=["willingness_to_pay_premium", "demand_for_value_added"],
            ),
            "standard": CustomerBehavior(
                segment="standard",
                price_sensitivity=0.6,
                service_preference=0.6,
                loyalty_drivers=["price_stability", "service_consistency", "relationship"],
                switching_costs=0.5,
                growth_indicators=["volume_discounts", "service_improvements"],
            ),
            "budget": CustomerBehavior(
                segment="budget",
                price_sensitivity=0.9,
                service_preference=0.3,
                loyalty_drivers=["low_price", "basic_reliability"],
                switching_costs=0.2,
                growth_indicators=["price_promotions", "cost_efficiency"],
            ),
        }

        for shipping_line in self.shipping_lines:
            if shipping_line.segment not in behavior_profiles:
                raise ValueError(
                    f"No behavior profile is defined for segment "
                    f"{shipping_line.segment!r}."
                )
            self.customer_behaviors[shipping_line.id] = behavior_profiles[
                shipping_line.segment
            ]

    def _initialize_competitive_landscape(self) -> None:
        """Store descriptive competition categories and intensity assumptions."""

        self.competitive_landscape = {
            "market_leaders": [],
            "price_warriors": [],
            "service_specialists": [],
            "niche_players": [],
            "market_challengers": [],
            "intensity_analysis": {
                "price_competition": 0.7,
                "service_competition": 0.6,
                "innovation_competition": 0.4,
                "partnership_competition": 0.5,
            },
        }

    def analyze_market_opportunities(self) -> List[MarketOpportunity]:
        """Return the illustrative opportunity descriptions used in this example."""

        opportunities = [
            MarketOpportunity(
                opportunity_type="emerging_trade_lanes",
                target_customers=["growth_focused_lines", "regional_expansion"],
                market_size=50000,
                growth_rate=0.25,
                profit_potential=0.18,
                competitive_advantage="first_mover_advantage",
                required_capabilities=["flexible_infrastructure", "customs_efficiency"],
                time_to_market=12,
                confidence_level=0.8,
            ),
            MarketOpportunity(
                opportunity_type="premium_services",
                target_customers=["high_value_lines", "time_sensitive_cargo"],
                market_size=30000,
                growth_rate=0.15,
                profit_potential=0.25,
                competitive_advantage="service_excellence",
                required_capabilities=["premium_facilities", "dedicated_staff"],
                time_to_market=8,
                confidence_level=0.9,
            ),
            MarketOpportunity(
                opportunity_type="partnership_ecosystem",
                target_customers=["logistics_providers", "integrated_shippers"],
                market_size=40000,
                growth_rate=0.20,
                profit_potential=0.12,
                competitive_advantage="network_effects",
                required_capabilities=["digital_platform", "partner_management"],
                time_to_market=16,
                confidence_level=0.7,
            ),
        ]
        self.identified_opportunities = opportunities
        return opportunities

    def design_strategic_intervention(
        self, opportunity: MarketOpportunity
    ) -> MarketIntervention:
        """Map a known opportunity category to a proposed intervention.

        The mapping specifies an action and candidate measures, not a forecast.
        Costs and expected impacts remain unspecified because no estimates for
        them are supplied in the example.
        """

        intervention_designs = {
            "emerging_trade_lanes": (
                MarketInterventionType.VOLUME_DISCOUNT,
                "Offer a volume discount for qualifying emerging trade lanes.",
                ["new_lane_volume", "customer_acquisition", "contribution_margin"],
                ["discount cannibalization", "uncertain lane demand"],
            ),
            "premium_services": (
                MarketInterventionType.PREMIUM_SERVICE,
                "Offer differentiated services for high-value and time-sensitive cargo.",
                ["premium_service_adoption", "service_reliability", "contribution_margin"],
                ["service-capacity constraints", "delivery-quality risk"],
            ),
            "partnership_ecosystem": (
                MarketInterventionType.PARTNERSHIP_PROGRAM,
                "Develop a partnership program with logistics providers and shippers.",
                ["active_partners", "partner-generated_volume", "joint_value_created"],
                ["partner coordination", "unclear value sharing"],
            ),
        }

        if opportunity.opportunity_type not in intervention_designs:
            raise ValueError(
                f"No intervention design is defined for opportunity type "
                f"{opportunity.opportunity_type!r}."
            )

        intervention_type, pricing_strategy, success_metrics, risk_factors = (
            intervention_designs[opportunity.opportunity_type]
        )
        return MarketIntervention(
            intervention_type=intervention_type,
            target_segments=list(opportunity.target_customers),
            pricing_strategy=pricing_strategy,
            expected_impact={},
            duration_weeks=opportunity.time_to_market,
            investment_required=None,
            success_metrics=success_metrics,
            risk_factors=risk_factors,
        )
```

```python
def design_strategic_interventions(
    market_shaper: MarketShapingEngine,
    opportunities: List[MarketOpportunity],
) -> List[MarketIntervention]:
    """Design one intervention for each supplied opportunity."""

    return [
        market_shaper.design_strategic_intervention(opportunity)
        for opportunity in opportunities
    ]
```

## 1. Define the example instance

The port and shipping-line records provide context for the planning discussion. Their capacities, costs, and demand values are not inputs to an optimization calculation in this notebook. The segment profiles describe assumed customer characteristics; they should not be interpreted as measured elasticities or causal effects.

```python
def create_concrete_example() -> Tuple[List[Port], List[ShippingLine], MarketParameters]:
    """Create the illustrative instance with three ports and three shipping lines."""

    ports = [
        Port(
            id=0,
            name="Port A",
            capacity=20000,
            costs=[80, 85, 82],
            strategic_position="leader",
        ),
        Port(
            id=1,
            name="Port B",
            capacity=18000,
            costs=[85, 78, 88],
            strategic_position="challenger",
        ),
        Port(
            id=2,
            name="Port C",
            capacity=22000,
            costs=[75, 90, 80],
            strategic_position="follower",
        ),
    ]

    shipping_lines = [
        ShippingLine(
            id=0,
            name="Global Shipping Line",
            demand=12000,
            priority=1,
            segment="premium",
            price_elasticity=-0.3,
            growth_potential=0.8,
            loyalty_score=0.9,
        ),
        ShippingLine(
            id=1,
            name="Pacific Trade Line",
            demand=8000,
            priority=2,
            segment="standard",
            price_elasticity=-0.6,
            growth_potential=0.6,
            loyalty_score=0.7,
        ),
        ShippingLine(
            id=2,
            name="Regional Container Line",
            demand=5000,
            priority=3,
            segment="budget",
            price_elasticity=-0.9,
            growth_potential=0.4,
            loyalty_score=0.5,
        ),
    ]

    market_params = MarketParameters(
        price_sensitivity=0.01,
        min_price=80,
        max_price=150,
        market_growth_rate=0.12,
        competitive_intensity=0.7,
    )

    return ports, shipping_lines, market_params


ports, shipping_lines, market_params = create_concrete_example()
market_shaper = MarketShapingEngine(ports, shipping_lines, market_params)

print("Proactive Demand and Market-Shaping Configuration")
print(f"Number of ports: {len(ports)}")
print(f"Number of shipping lines: {len(shipping_lines)}")
print(f"Total shipping-line demand: {sum(line.demand for line in shipping_lines):,} TEU")
print(f"Total port capacity: {sum(port.capacity for port in ports):,} TEU")
print(f"Market growth-rate assumption: {market_params.market_growth_rate:.1%}")
print(f"Competitive-intensity assumption: {market_params.competitive_intensity:.1%}")

print("\nPort strategic positions:")
for port in ports:
    print(f"- {port.name}: {port.strategic_position.title()}")

print("\nShipping-line segments:")
for line in shipping_lines:
    print(
        f"- {line.name}: {line.segment.title()}, "
        f"growth potential {line.growth_potential:.1%}, "
        f"loyalty score {line.loyalty_score:.1%}"
    )

print("\nDescriptive customer behavior profiles:")
for line in shipping_lines:
    behavior = market_shaper.customer_behaviors[line.id]
    print(
        f"- {line.name}: price-sensitivity profile "
        f"{behavior.price_sensitivity:.1f}, service-preference profile "
        f"{behavior.service_preference:.1f}"
    )
```

## 2. Identify market opportunities

An opportunity record combines a target market, a market-size assumption, a growth-rate assumption, a potential margin, required capabilities, a time-to-market estimate, and a confidence level. These fields organize the planning discussion; they do not by themselves establish that an opportunity is feasible or profitable. The function below returns the illustrative records defined in the engine.

```python
def analyze_market_opportunities(
    market_shaper: MarketShapingEngine,
) -> List[MarketOpportunity]:
    """Display and return the opportunity descriptions in the example."""

    print("\n" + "=" * 80)
    print("MARKET OPPORTUNITY ANALYSIS")
    print("=" * 80)

    opportunities = market_shaper.analyze_market_opportunities()
    print(f"\nIllustrative opportunities: {len(opportunities)}")

    for index, opportunity in enumerate(opportunities, start=1):
        print(f"\n{index}. {opportunity.opportunity_type.replace('_', ' ').title()}")
        print(f"   Target customers: {', '.join(opportunity.target_customers)}")
        print(f"   Assumed market size: {opportunity.market_size:,} TEU")
        print(f"   Assumed growth rate: {opportunity.growth_rate:.1%}")
        print(f"   Potential margin assumption: {opportunity.profit_potential:.1%}")
        print(f"   Competitive advantage: {opportunity.competitive_advantage}")
        print(f"   Required capabilities: {', '.join(opportunity.required_capabilities)}")
        print(f"   Estimated time to market: {opportunity.time_to_market} weeks")
        print(f"   Confidence assumption: {opportunity.confidence_level:.1%}")

    return opportunities


opportunities = analyze_market_opportunities(market_shaper)
```

## 3. Design strategic interventions

The intervention design step is a rule-based association: each of the three supplied opportunity categories has a corresponding intervention type, descriptive pricing or partnership strategy, candidate success measures, and risks to consider. The duration is taken from that opportunity's time-to-market field. No numerical impact or investment estimate is assigned because the example supplies no evidence from which to calculate one.

```python
def implement_strategic_interventions(
    market_shaper: MarketShapingEngine,
    opportunities: List[MarketOpportunity],
) -> List[Dict[str, object]]:
    """Design interventions and report known details without forecasting impact."""

    print("\n" + "=" * 80)
    print("STRATEGIC INTERVENTION DESIGN")
    print("=" * 80)

    interventions = design_strategic_interventions(market_shaper, opportunities)
    intervention_results = []

    for index, (opportunity, intervention) in enumerate(
        zip(opportunities, interventions), start=1
    ):
        market_shaper.active_interventions.append(intervention)
        market_shaper.intervention_history.append(intervention)

        result = {
            "opportunity_type": opportunity.opportunity_type,
            "intervention_type": intervention.intervention_type.value,
            "target_segments": intervention.target_segments,
            "expected_impact": intervention.expected_impact,
            "investment": intervention.investment_required,
            "duration_weeks": intervention.duration_weeks,
        }
        intervention_results.append(result)

        print(f"\nIntervention {index}: {opportunity.opportunity_type.replace('_', ' ').title()}")
        print(f"Type: {intervention.intervention_type.value.replace('_', ' ').title()}")
        print(f"Target customers: {', '.join(intervention.target_segments)}")
        print(f"Strategy: {intervention.pricing_strategy}")
        print(f"Planning horizon: {intervention.duration_weeks} weeks")
        print("Investment estimate: not specified")
        print("Expected numerical impact: not estimated")
        print(f"Candidate success measures: {', '.join(intervention.success_metrics)}")
        print(f"Risks to assess: {', '.join(intervention.risk_factors)}")

    return intervention_results


intervention_results = implement_strategic_interventions(
    market_shaper,
    opportunities,
)
```

## 4. Assess what can be concluded

An impact assessment requires observed or credibly estimated effects, such as changes in demand, customer retention, market share, or revenue. The intervention records contain no numerical impact estimates, so aggregating them into a performance score would confuse missing information with a zero effect. The analysis therefore reports those outcomes as unavailable and identifies what evidence would be needed.

```python
def analyze_market_shaping_impact(
    intervention_results: List[Dict[str, object]],
) -> Dict[str, Optional[float]]:
    """Report impact measures only when numerical estimates are supplied."""

    print("\n" + "=" * 80)
    print("MARKET-SHAPING IMPACT ASSESSMENT")
    print("=" * 80)

    impact_summary: Dict[str, Optional[float]] = {
        "demand_shift_percentage": None,
        "behavior_change_score": None,
        "market_growth_percentage": None,
        "competitive_position_gain": None,
        "customer_loyalty_improvement": None,
        "overall_market_shaping_score": None,
    }

    has_numeric_impact = any(
        bool(result["expected_impact"])
        for result in intervention_results
    )

    if has_numeric_impact:
        print("Numerical impact entries are present; interpret each metric using its definition and units.")
    else:
        print("Numerical intervention impacts are not specified in the example.")

    for metric in impact_summary:
        print(f"{metric.replace('_', ' ').title()}: not estimated")

    print(
        "\nTo estimate realized impact, collect comparable baseline and follow-up "
        "measures and document the method used to attribute changes to each intervention."
    )
    return impact_summary


impact_summary = analyze_market_shaping_impact(intervention_results)
```

## 5. Visualize the planning information

The plots summarize the supplied opportunity assumptions and the intervention designs. They do not display achieved performance. In particular, an empty impact field means that an estimate was not supplied; it is not evidence of zero impact. The case figures shown in the final panel are contextual benchmarks, not measurements from this implementation.

```python
def visualize_market_shaping(
    opportunities: List[MarketOpportunity],
    intervention_results: List[Dict[str, object]],
    impact_summary: Dict[str, Optional[float]],
):
    """Plot opportunity assumptions and clearly mark unavailable outcomes."""

    fig, axes = plt.subplots(2, 3, figsize=(18, 12))
    fig.suptitle("Proactive Demand and Market-Shaping Planning", fontsize=16, fontweight="bold")

    opportunity_names = [
        opportunity.opportunity_type.replace("_", " ").title()
        for opportunity in opportunities
    ]

    ax = axes[0, 0]
    market_sizes = [opportunity.market_size / 1000 for opportunity in opportunities]
    x = np.arange(len(opportunity_names))
    ax.bar(x, market_sizes, alpha=0.75, label="Assumed market size")
    ax.set_title("Opportunity Market-Size Assumptions")
    ax.set_ylabel("Market size (thousand TEU)")
    ax.set_xticks(x)
    ax.set_xticklabels(opportunity_names, rotation=35, ha="right")
    ax.legend()
    ax.grid(True, axis="y", alpha=0.3)

    ax = axes[0, 1]
    growth_rates = [opportunity.growth_rate * 100 for opportunity in opportunities]
    confidence_levels = [opportunity.confidence_level * 100 for opportunity in opportunities]
    width = 0.35
    ax.bar(x - width / 2, growth_rates, width, label="Assumed growth rate (%)")
    ax.bar(x + width / 2, confidence_levels, width, label="Confidence assumption (%)")
    ax.set_title("Opportunity Assumptions")
    ax.set_ylabel("Percentage (%)")
    ax.set_xticks(x)
    ax.set_xticklabels(opportunity_names, rotation=35, ha="right")
    ax.legend()
    ax.grid(True, axis="y", alpha=0.3)

    ax = axes[0, 2]
    ax.axis("off")
    ax.set_title("Investment and Expected Impact")
    ax.text(
        0.05,
        0.75,
        "Investment estimates: not supplied\n\n"
        "Numerical intervention impacts: not estimated\n\n"
        "These unknowns cannot be interpreted as zero.",
        transform=ax.transAxes,
        va="top",
        fontsize=12,
    )

    ax = axes[1, 0]
    segment_counts = {}
    for result in intervention_results:
        for segment in result["target_segments"]:
            segment_counts[segment] = segment_counts.get(segment, 0) + 1
    if segment_counts:
        ax.bar(segment_counts.keys(), segment_counts.values(), alpha=0.75)
        ax.tick_params(axis="x", rotation=35)
    else:
        ax.text(0.5, 0.5, "No target customers specified", ha="center", va="center")
    ax.set_title("Intervention Target-Customer Mentions")
    ax.set_ylabel("Number of intervention designs")
    ax.grid(True, axis="y", alpha=0.3)

    ax = axes[1, 1]
    intervention_types = [
        str(result["intervention_type"]).replace("_", " ").title()
        for result in intervention_results
    ]
    type_counts = {}
    for intervention_type in intervention_types:
        type_counts[intervention_type] = type_counts.get(intervention_type, 0) + 1
    if type_counts:
        ax.bar(type_counts.keys(), type_counts.values(), alpha=0.75)
        ax.tick_params(axis="x", rotation=35)
    else:
        ax.text(0.5, 0.5, "No interventions designed", ha="center", va="center")
    ax.set_title("Designed Intervention Types")
    ax.set_ylabel("Count")
    ax.grid(True, axis="y", alpha=0.3)

    ax = axes[1, 2]
    benchmark_names = ["New-segment growth", "Customer loyalty", "Market position"]
    benchmark_values = [25, 18, 32]
    ax.bar(benchmark_names, benchmark_values, alpha=0.75, label="Case-description benchmark")
    ax.set_title("Contextual Case Figures")
    ax.set_ylabel("Reported percentage (%)")
    ax.tick_params(axis="x", rotation=30)
    ax.legend()
    ax.grid(True, axis="y", alpha=0.3)
    ax.text(
        0.5,
        0.95,
        "Not calculated by this notebook",
        transform=ax.transAxes,
        ha="center",
        va="top",
        fontsize=9,
    )

    fig.tight_layout(rect=(0, 0, 1, 0.95))
    plt.show()
    return fig


visualization_fig = visualize_market_shaping(
    opportunities,
    intervention_results,
    impact_summary,
)
```

## 6. Interpretation and limitations

The example illustrates a progression from describing market opportunities to selecting plausible intervention concepts. It does not demonstrate that an intervention causes demand growth, customer loyalty, or a stronger competitive position. Those outcomes require operational definitions, data, and an evaluation design.

The supplied price-sensitivity values and market parameters are not connected to a demand equation, price decision, or revenue calculation here. Consequently, the notebook makes no quantitative pricing recommendation. Likewise, the opportunity market sizes, growth rates, potential margins, confidence levels, and time-to-market values are planning assumptions rather than validated forecasts.

A practical assessment would define measurable outcomes before implementation, record baseline values, track intervention costs and follow-up outcomes, and assess whether observed changes are plausibly attributable to the intervention. Any resulting estimates should state their units, time horizon, uncertainty, and assumptions.

### Case-description context

The accompanying case description presents a three-part strategy: volume discounts for emerging trade lanes, premium services for high-value customers, and partnership programs with logistics providers. It reports 25% growth in new market segments, an 18% increase in customer loyalty, and a 32% improvement in market position. These figures are reported case claims, not outputs of the illustrative data or code in this notebook.

### When this framework is useful

This structured planning framework is useful when a port is exploring emerging markets, differentiated services, or partnership opportunities and needs to make assumptions and evidence gaps explicit. It is not sufficient by itself for choosing prices, estimating causal effects, or determining whether an investment is profitable.
